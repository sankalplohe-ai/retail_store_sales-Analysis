```python
import pandas as pd
import numpy as np
df=pd.read_csv('retail_store_sales project 1.csv')
print("first 5 rows:")
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
df.columns=df.columns.str.strip().str.lower().str.replace(' ','_')
print("updated columns:")
print(df.columns)
#missing values
print(df.isnull().sum())
#handling missing values
df['item'].fillna('unknown',inplace=True)
df['discount_applied'].fillna('False',inplace=True)
df['price_per_unit'].fillna(df['price_per_unit'].median(),inplace=True)
df['quantity'].fillna(df['quantity'].median(),inplace=True)
df['total_spent'].fillna(df['total_spent'].median(),inplace=True)
#removing duplicate rows
print("Duplicate Rows:",df.duplicated().sum())
df.drop_duplicates(inplace=True)
#correcting data type
df['transaction_date']=pd.to_datetime(df['transaction_date'])
df['category']=df['category'].astype(str).str.strip().str.lower()
df['payment_method']=df['price_per_unit'].astype(str).str.strip().str.lower()
df['location']=df['location'].astype(str).str.strip().str.lower()
#cleaned data
print(df.isnull().sum())
#outlier detection
Q1=df['total_spent'].quantile(0.25)
Q3=df['total_spent'].quantile(0.75)
IQR=Q3-Q1
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
df=df[(df['total_spent']>=lower_limit) & (df['total_spent']<=upper_limit)]
print("Dataset shape After Outlier Removal:",df.shape)
# DATA ANALYSIS
total_revenue=df['total_spent'].sum()
print("Total Revenue:",total_revenue)
category_sales=df.groupby('category')['total_spent'].sum().sort_values(ascending=False)
print(category_sales)
# payment Method Analysis
payment_sales=df.groupby('payment_method')['total_spent'].sum().sort_values(ascending=False)
print(payment_sales)
# Location-wise sales analysis
location_sales=df.groupby('location')['total_spent'].sum().sort_values(ascending=False)
print(location_sales)
# Monthly sales trend analysis
df['month']=df['transaction_date'].dt.month
monthly_sales=df.groupby('month')['total_spent'].sum().sort_values(ascending=False)
print(monthly_sales)
#top-selling items
item_sales=df.groupby('item')['total_spent'].sum().sort_values(ascending=False)
print(item_sales)
```
